## Popcorn Hack 2

```python

class Player {
  constructor(name, health) {
    this.name = name;
    this.health = health;
  }

  // This method is already complete. Use it as a model.
  heal() {
    this.health += 10;
  }

  // TODO 1: Create takeDamage() and subtract 10 from health.
  takeDamage() {
    this.health -= 10;
  }
}
let player = new Player("Alex", 50);
player.heal();
console.log("Health after healing: " + player.health); // 60

// TODO 2: Call takeDamage() and print "Health after damage: " with the final health.
player.takeDamage();
console.log("Health after damage: " + player.health); //50
```

## Homework Hack

```python

// Use the steps below as a guide for your own solution.
// 1. Define a class.
class Enemy {
// 2. Add a constructor with two properties.
    constructor(name, damageDealing) {
        this.name = name;
        this.damageDealing = damageDealing;
    }
// 3. Add one method that increases a value.
    increaseDamage() {
        this.damageDealing +=10
    }
// 4. Add one method that decreases a value.
    decreaseDamage() {
        this.damageDealing -=20
    }
}
// 5. Create an object with new and test both methods.
let enemy = new Enemy("Steve", 30);

enemy.increaseDamage();
console.log("How much damage " + enemy.name + " deals after increased power: " + enemy.damageDealing); // 40

enemy.decreaseDamage();
console.log("How much damage " + enemy.name + " deals after being hit by the player: " + enemy.damageDealing); // 20
```
